### 3. Ultimate Vector Space Search Matrix 🔍

```
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

text_embs = np.load('embeddings/raw/text_layer-1.npy')
image_embs = np.load('embeddings/raw/image_layer-1.npy')
audio_embs = np.load('embeddings/raw/audio_layer-1.npy')

print('\n--- 🔊 AUDIO CLIP QUERY --> [TEXT DB] & [IMAGE DB] ---')
text_sims = cosine_similarity(audio_embs[0].reshape(1, -1), text_embs)[0]
img_sims = cosine_similarity(audio_embs[0].reshape(1, -1), image_embs)[0]

top_text = np.argsort(text_sims)[::-1][0]
top_img = np.argsort(img_sims)[::-1][0]

print(f'⭐ Best Semantic Text Match # {top_text} (Score: {text_sims[top_text]:.3f})')
print(f'⭐ Best Semantic Image Match # {top_img} (Score: {img_sims[top_img]:.3f})')
```
