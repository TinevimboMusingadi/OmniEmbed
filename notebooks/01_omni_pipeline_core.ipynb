## 4. 🗜️ Information Compression Mapping
See exactly how efficient the backbone's dimensionality arrays handle aggressive PCA limits.

````
import numpy as np
import matplotlib.pyplot as plt
from omniembed.compression import PCACompressor

t_embs = np.load('embeddings/raw/text_layer-1.npy')
dims = [10, 64, 256, 512, 1024, 2048]
variances = [PCACompressor(target_dim=d).fit(t_embs) for d in dims]

plt.figure(figsize=(9, 5))
plt.plot(dims, variances, marker='o', color='purple', linewidth=2)
plt.axvline(512, color='red', linestyle='--', label='Sweet Spot (512d)')
plt.title('Accumulated Variance', fontsize=14)
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()
````
