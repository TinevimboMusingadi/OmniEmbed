### Plotting PCA Explained Variance 📉
See exactly how efficiently the backbone compresses data!

```python
import numpy as np
import matplotlib.pyplot as plt
from omniembed.compression import PCACompressor

t_embs = np.load('embeddings/raw/text_layer-1.npy')

dims = [10, 32, 64, 128, 256, 512, 1024, 2048]
variances = []

print('Iterative dimensionality reduction calculating target PCA Thresholds...')
for d in dims:
    comp = PCACompressor(target_dim=d)
    var = comp.fit(t_embs)
    variances.append(var)

plt.figure(figsize=(9, 5))
plt.plot(dims, variances, marker='o', linestyle='-', color='purple', linewidth=2)
plt.axvline(512, color='red', linestyle='--', label='Production Target Sweet Spot (512d)')
plt.title('Accumulated Variance Maintained vs Database Storage Dimensions', fontsize=14)
plt.xlabel('Dimensions Kept', fontsize=12)
plt.ylabel('Cumulative Variance', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
