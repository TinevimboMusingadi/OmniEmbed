### 📊 Visualize the Geometry Corrections natively out of the box

```
import numpy as np
import matplotlib.pyplot as plt
from omniembed.viz import plot_modality_umap

# Raw Tensors
raw_t = np.load('embeddings/raw/text_layer-1.npy')
raw_i = np.load('embeddings/raw/image_layer-1.npy')

# Aligned Space Tensors
align_t = np.load('embeddings/aligned/whitened/text_aligned.npy')
align_i = np.load('embeddings/aligned/whitened/image_aligned.npy')

print('\nPlotting Raw Embeddings (Notice the huge Modality Gap separating the clusters!):')
plot_modality_umap({'Text': raw_t, 'Image': raw_i}, title='Raw Layer -1 Separation Gap')
print('\nPlotting completely Aligned Embeddings (Merged smoothly together):')
plot_modality_umap({'Text': align_t, 'Image': align_i}, title='ZCA Whitened Unified Space')
```
