### 3. Ultimate Cross-Modal Universal Algorithm! 🔍

```
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Load the Multi-modal databases into RAM
text_embs = np.load('embeddings/raw/text_layer-1.npy')
image_embs = np.load('embeddings/raw/image_layer-1.npy')
audio_embs = np.load('embeddings/raw/audio_layer-1.npy')

# Setup universal search taking the Audio vector we heard earlier
query_audio = audio_embs[0].reshape(1, -1)

print('--- 🔊 EXECUTING: AUDIO QUERY #0 SEARCHING -> [TEXT DB] & [IMAGE DB] ---\n')
# Massive Dot-Product Cosine Similarity 
text_sims = cosine_similarity(query_audio, text_embs)[0]
img_sims = cosine_similarity(query_audio, image_embs)[0]

# Retrieve Top Hits
top_text = np.argsort(text_sims)[::-1][0]
top_img = np.argsort(img_sims)[::-1][0]

print(f'⭐ Best Semantic Text Match: Document #{top_text} (Cosine Score: {text_sims[top_text]:.3f})')
print(f'⭐ Best Semantic Image Match: Image #{top_img} (Cosine Score: {img_sims[top_img]:.3f})')
```
